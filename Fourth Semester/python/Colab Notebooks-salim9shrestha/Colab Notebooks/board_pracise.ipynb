```python
#wap to take list of numbers and return the maximum and minimum values.
def find_max_min(num):
  maximum = num[0]
  minimum = num[0]

  for number in num:
    if number>maximum:
      maximum = number

    elif number<minimum:
      minimum = number

  return maximum, minimum
```

```python
num = [10, 20, 30, 40, 5];
max, min = find_max_min(num)

print("maximum=", max)
print("minimum=", min)

```

```python
class register():
  def __init__(self,username,password):
    self.username=username
    self.password=password

  def w(self):
    with open("readme.txt","w") as file:
      file.write("i write \n")

  def r(self):
    with open("readme.txt","r") as file:
      print(file.read())

  def a(self):
    with open("readme.txt","a") as file:
      file.write("i append \n")

  def login(self):
    print("you are login")

class student(register):
  def __init__(self,username,password,rollno):
    super().__init__(username,password)
    self.rollno=rollno

  def enroll(self):
    print("enrolling")

class instructor(register):
  def __init__(self,username,password,id):
    super().__init__(username,password)
    self.id=id


  def check(self):
    print("checked")
```

```python
s=student("salim","salim@1234",20)
i=instructor("subik","subik@1234",1)

s.enroll()
print(s.username)
print(s.rollno)
print(i.id)

i.check()

s.login()
s.w()
s.r()
s.a()
s.r()
```

```python
#to find minimum and maximum values function

def findmm(numbers):
  min=numbers[0]
  max=numbers[0]

  for i in numbers:
    if i>max:
      max=i
    if i<min:
      min=i

  return min, max
```

```python
numbers=[3,1,4,0,1]

min, max =findmm(numbers)

print(max)
print(min)
```
