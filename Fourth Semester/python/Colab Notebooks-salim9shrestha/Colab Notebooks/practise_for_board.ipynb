```python
def findmm(num):
    maxx = num[0]
    minn = num[0]

    for number in num:
        if number > maxx:
            maxx = number
        elif number < minn:
            minn = number

    return maxx, minn
```

```python
num = [20, 30, 40, 5]

maxx, minn = findmm(num)

print(f"maximum no {maxx}")
print(f"minimum no {minn}")
```

```python
#maximum and minimum values
def findnum(numbers):
  max =numbers[0]
  min =numbers[0]

  for num in numbers:
    if num < max:
      max=num
    if num > min:
      min=num

  return max, min

```

```python
numbers= [10,20,30,40]

max, min= findnum(numbers)

print(max)
print(min)
```
