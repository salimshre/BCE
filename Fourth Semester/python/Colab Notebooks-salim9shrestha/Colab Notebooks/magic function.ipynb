```python
class fraction:
  def __init__(self,n,d):
    self.num=n
    self.den=d

  def __str__(self):
    return '{}/{}'.format(self.num,self.den)

  def __add__(see, am):
    new_num = see.num*am.den + see.den*am.num
    new_den = see.den*am.den
    return '{}/{}'.format(new_num,new_den)
```

```python
obj1 =fraction(2,2)
obj2 =fraction(4,1)

print(obj1)
print(obj2)
print(obj1+obj2)
```
